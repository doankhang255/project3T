```python
import os
import requests
import pandas as pd

your_key = os.getenv("X_API_KEY")
url = "https://gateway.datacore.vn/data/ds/search"

headers = {
    "x-api-key": "",
    "Content-Type": "application/json"
}

payload = {
    "dataSetCode": "dataset_news", #name of datasetCode
    "conditions": [],
    "selectFields": [],
    "page": 1, 
    "limit": 10000 # 719603
}
res = requests.post(url, headers=headers, json=payload)
# print(res)
data = res.json()
print(data)

rows = data["data"]["dataDetail"]
cols = data["data"]["fields"]

print("Số dòng request trả về:", len(rows))
print("Số cột:", len(cols))

# df = pd.DataFrame(
#     data["data"]["dataDetail"],
#     columns=data["data"]["fields"]x
# )
# df.head(1000)
```
