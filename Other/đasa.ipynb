```python
import pandas as pd

file_path = "data_news_merged.parquet"

file_path1 = "data_news_merged_1.parquet"

df = pd.read_parquet(file_path)

before = len(df)

df = df.drop_duplicates(
    subset=["id"],
    keep="first"
)

after = len(df)

df.to_parquet(
    file_path1,
    index=False,
    compression="zstd"
)

print(f"Trước khi xóa: {before:,} dòng")
print(f"Sau khi xóa: {after:,} dòng")
print(f"Đã xóa: {before - after:,} dòng duplicate")
```

```python
duplicate_id = df[df.duplicated(subset=["id"], keep=False)]

print("Số dòng duplicate theo id:", len(duplicate_id))
# duplicate_id.sort_values("id")
```
